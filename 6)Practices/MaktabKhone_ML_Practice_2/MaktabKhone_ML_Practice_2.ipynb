### Import Required Libraries

Import NumPy for numerical arrays, pandas for tabular data, and DecisionTreeClassifier for the first classification model.

```python
# Import NumPy for numerical arrays, pandas for tabular data, and DecisionTreeClassifier for the first classification model.
import numpy as np
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
```

### Load and Inspect the Dataset

Load the heart-disease dataset from a CSV file and display the first 20 rows to inspect the data.

```python
# Load the heart-disease dataset from a CSV file and display the first 20 rows to inspect the data.
my_data = pd.read_csv("heart.csv")
my_data[:20]
```

### Check Dataset Dimensions

Show the number of rows and columns in the dataset.

```python
# Show the number of rows and columns in the dataset.
my_data.shape
```

```python
my_data.describe()
```

### Select Input Features

Select the input features used to predict heart disease and display the first 20 rows.

```python
# Select the input features used to predict heart disease and display the first 20 rows.
X = my_data[['age', 'sex', 'cp', 'trtbps', 'chol', 'fbs', 'restecg', 'thalachh', 'exng', 'caa']]
X[:20]
```

### Select the Target Variable

Select the target column (`output`), where 1 usually indicates heart disease and 0 indicates no heart disease.

```python
# Select the target column (`output`), where 1 usually indicates heart disease and 0 indicates no heart disease.
Y = my_data[['output']]
Y[:20]
```

### Import Train-Test Split Tool

Import `train_test_split` to divide the dataset into training and testing subsets.

```python
# Import `train_test_split` to divide the dataset into training and testing subsets.
from sklearn.model_selection import train_test_split
```

### Split Data into Training and Test Sets

Split features and target into 70% training data and 30% test data. `random_state=3` makes the split reproducible.

```python
# Split features and target into 70% training data and 30% test data. `random_state=3` makes the split reproducible.
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.3, random_state=3)
```

### Verify Training and Test Set Sizes

Print the dimensions of the training and test sets to verify the split.

```python
# Print the dimensions of the training and test sets to verify the split.
print(f"X_train size is: {X_train.shape}")
print(f"Y_train size is: {Y_train.shape}")
print(f"X_test size is: {X_test.shape}")
print(f"Y_test size is: {Y_test.shape}")
```

### Train and Evaluate Decision Tree Models

Train Decision Tree models using different splitting criteria and maximum depths, then measure accuracy on the test set.

```python
# Train Decision Tree models using different splitting criteria and maximum depths, then measure accuracy on the test set.
from sklearn import metrics
citerions    = ["gini", "entropy", "log_loss"]
depths       = [None, 4, 6]
test_results = {}

for citerion in citerions:
    test_results[citerion] = {}
    for depth in depths:
        treemodel = DecisionTreeClassifier(criterion=citerion, max_depth=depth, random_state=1, splitter="random")
        treemodel.fit(X_train, Y_train)
        predicted_tree = treemodel.predict(X_test)
        score = metrics.accuracy_score(Y_test, predicted_tree)
        test_results[citerion][depth] = (score * 100)
```

### Display Decision Tree Accuracy Results

Display the test accuracy for every combination of decision-tree criterion and maximum depth. base on the results, as you can see we achive the accuracy of 87.91% via the "Entropy" citerion and max_depth of 6.

```python
# Display the test accuracy for every combination of decision-tree criterion and maximum depth.
for citerion_item in list(test_results.keys()):
    print((10*"#") + f"{citerion_item}" + (10*"#"))
    for depth_item in (test_results[citerion_item].keys()):
        print(f"the {citerion_item} DecisionTree with depth of {depth_item} Accuracy is: {test_results[citerion_item][depth_item]:.2f}%")
    print()
```

Let's build a SVM model for our "heart.csv" datasets.

### Visualize Cholesterol and Blood Pressure

### Inspect Column Data Types

Inspect the data type of every column before preprocessing.

### Convert `oldpeak` to Integer Type

Convert the continuous-looking `oldpeak` column to integer type, then display the updated column types.

```python
# Convert the continuous-looking `oldpeak` column to integer type, then display the updated column types.
my_data['oldpeak'] = my_data['oldpeak'].astype('int')
my_data.dtypes
```

### Prepare Features for the SVM Model

Prepare the feature matrix for the SVM model by selecting the same features and converting them to a NumPy array.

```python
# Prepare the feature matrix for the SVM model by selecting the same features and converting them to a NumPy array.
feature_my_data = my_data[['age', 'sex', 'cp', 'trtbps', 'chol', 'fbs', 'restecg', 'thalachh', 'exng', 'caa']]
X_SVM = np.array(feature_my_data)
X_SVM[:5]
```

### Prepare Target Labels for the SVM Model

Convert the target labels to integers and store them as a NumPy array for scikit-learn.

```python
# Convert the target labels to integers and store them as a NumPy array for scikit-learn.
my_data['output'] = my_data['output'].astype('int')
Y_SVM = np.array(my_data['output'])
Y_SVM[:5]
```

### Split SVM Data into Training and Test Sets

Split the NumPy feature matrix and target array into training and test sets, using 30% of the data for testing.

```python
# Split the NumPy feature matrix and target array into training and test sets, using 30% of the data for testing.
Xtrain, Xtest, Ytrain, Ytest = train_test_split(X_SVM, Y_SVM, test_size=0.3, random_state=4)
print ('Train set:', Xtrain.shape,  Ytrain.shape)
print ('Test set:', Xtest.shape,  Ytest.shape)
```

### Train and Evaluate SVM Models

Train Support Vector Machine classifiers with four different kernels and store each model's classification report.

```python
# Train Support Vector Machine classifiers with four different kernels and store each model's classification report.
from sklearn import svm
from sklearn.metrics import classification_report

kernels  = ['linear', 'poly', 'rbf', 'sigmoid']
result   = {}
for kernel in kernels:
    svm_model = svm.SVC(kernel=kernel)
    svm_model.fit(Xtrain, Ytrain)
    predicted_data = svm_model.predict(Xtest)
    result[kernel] = classification_report(Ytest, predicted_data)

print(result)
```

### Display SVM Classification Reports

Print the classification report for each SVM kernel, including precision, recall, F1-score, and accuracy.

```python
# Print the classification report for each SVM kernel, including precision, recall, F1-score, and accuracy.
for item in list(result.keys()):
    print((10*"--") + f"classification report of {item}" + (10*"--"))
    print(result[item])
```
